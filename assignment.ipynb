#### Question 3: Hedging with a misspecified volatility

A bank sells $N = 1{,}000$ European calls with strike $K_1 = 100$ and maturity $T_1 = 0.5$. It prices and hedges them with the Black–Scholes model, using its estimate $\sigma_{\text{model}} = 0.20$ of the volatility. Let $F(t, s; \sigma)$ denote the Black–Scholes price of one call at time $t$ when the stock price is $s$ and the volatility is $\sigma$; the premium the bank receives is $N \cdot F(0, S_0; \sigma_{\text{model}})$. In reality the stock follows a geometric Brownian motion with drift $\mu = 0.08$ and true volatility $\sigma_{\text{true}}$, which may differ from $\sigma_{\text{model}}$. Take $S_0 = 100$ and $r = 0.03$.

**Delta hedge.** The bank rebalances on the trading days $t_k = k/252$, $k = 0, 1, \dots, 125$. At $t_k$ it holds

$$\phi_{t_k} = N \cdot \frac{\partial F}{\partial s}(t_k, S_{t_k}; \sigma_{\text{model}})$$

shares. This is the hedge ratio: the number of shares that makes the value of the total position (short calls plus shares) insensitive to small changes in the stock price. The bank keeps this number of shares until $t_{k+1}$. The money-market account finances the trades, so that the strategy is self-financing:

$$\psi_0 B_0 = N \cdot F(0, S_0; \sigma_{\text{model}}) - \phi_0 S_0, \qquad \psi_{t_k} B_{t_k} = \psi_{t_{k-1}} B_{t_k} - (\phi_{t_k} - \phi_{t_{k-1}})\, S_{t_k}.$$

**(a)** Simulate at least 10,000 scenarios for each of $\sigma_{\text{true}} = 0.15$, $0.20$ and $0.25$. Report the mean, standard deviation and the 5% and 95% quantiles of the discounted P&L.

One can show (cf. the gamma–theta P&L on the slides) that

$$d\Pi_t - r\Pi_t\,dt = \tfrac12 N \cdot \Gamma_t S_t^2 \left(\sigma_{\text{model}}^2 - \sigma_{\text{true}}^2\right) dt.$$

**(b)(ii)** Evaluate the discounted, integrated right-hand side, $\int_0^{T_1} e^{-rt}\, \tfrac12 N \cdot \Gamma_t S_t^2 \left(\sigma_{\text{model}}^2 - \sigma_{\text{true}}^2\right) dt$, along your simulated paths and compare its mean with the mean P&L from (a).
